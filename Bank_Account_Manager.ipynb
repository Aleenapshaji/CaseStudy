#***1. Bank Account Manager***
***a. Simulate a bank account with deposit, withdrawal, and balance
enquiry operations
b. Use variables and data types to store account information
c. Implement conditional statements to check for sufficient balance
before withdrawal
d. Use functions to perform deposit, withdrawal, and balance inquiry
operations***

```python
balance = 0.0
def menu():                             # Sign-in Menu
  status = True
  while status:
    print('======MENU=======')
    print('1. Deposit')
    print('2. Withdrawal')
    print('3. Check Balance')
    print('4. Exit')

    user_choice = input("Enter your choice : ")
    if(user_choice == '1'):
      deposit()
    elif(user_choice == "2"):
      withdrawal()
    elif(user_choice == '3'):
      check_balance()
    elif(user_choice == '4'):
      print("Thankyou")
      status = False
      break
    else:
      print("Please enter a valid choice : ")



def deposit():                              # Deposit Function
  global balance
  amount_deposited = float(input("Enter the amount to be deposited : "))
  balance+= amount_deposited
  print(f"{amount_deposited} deposited successfully")

def withdrawal():                           # Withdrawal Function
  global balance
  amount_withdrawed = float(input("Enter amount to be withdrawed : "))
  if(amount_withdrawed <= balance):
    balance -= amount_withdrawed
    print(f"Current Balance : {balance} ")
  else :
    print("Insufficient Balance !!!")



def check_balance():                        # Check balance Function
  print(f"Balance : {balance}")


def main():                                 # Main Function
  print('Sign in')
  account_holder_name = input("Enter your Account holder's name :")
  account_number = int(input("Enter the Account number : "))
  menu()


if __name__ == '__main__':
  main()

```
